```python
from pathlib import Path
import random, shutil, zipfile

# ---- edit these ----
SRC = Path("/Users/samswitz/GitHub/micro-research/diffusion/data/celeba/img_align_celeba")
N   = 1000
DST = Path(f"/Users/samswitz/GitHub/micro-research/diffusion/data/celeba_subsets/subset_{N}")
SEED = 0
# --------------------

DST.mkdir(parents=True, exist_ok=True)

imgs = sorted(SRC.glob("*.jpg"))  # only JPGs; avoids copying .DS_Store, etc.
if len(imgs) < N:
    raise ValueError(f"Requested {N} but only found {len(imgs)} images")

random.seed(SEED)
subset = random.sample(imgs, N)

for p in subset:
    shutil.copy2(p, DST / p.name)

# Zip for upload to Colab
zip_path = DST.with_suffix(".zip")
with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for p in DST.iterdir():
        if p.is_file():
            z.write(p, arcname=p.name)

print(f"Subset folder: {DST}")
print(f"Zipped file: {zip_path}")
```
