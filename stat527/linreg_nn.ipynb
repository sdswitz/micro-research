```python
X = torch.tensor(df.drop(columns=['lpsa']).values, dtype=torch.float32)
y = torch.tensor(df['lpsa'].values, dtype=torch.float32).view(-1, 1)

print(X.shape, y.shape)
```

```python
class LinReg(nn.Module):
    """
    Linear Regression model using pytorch with MSE loss and backpropagation

    Args:
        nn (_type_): _description_
    """
    def __init__(self, in_features: int, out_features: int):
        super(LinReg, self).__init__()
        self.linear = nn.Linear(in_features=in_features, out_features=out_features)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.linear(x)
    
model = LinReg(in_features=p, out_features=1)

pred = model(X)
loss = F.mse_loss(pred, y)
print(loss)
```
