## Basic OLS regression

```python
class LinearModel:
    
    def __init__(
                self,
                X: np.ndarray,
                y: np.ndarray,
                coefficients: np.ndarray,
                fitted_values: np.ndarray
            ):
        self.X = X,
        self.y = y,
        self.coefficients = coefficients,
        self.fitted_values = fitted_values
        
        self.n = len(y)
```

```python
def least_squares(X, y, intercept=True):
    assert X.shape[0] > X.shape[1]
    assert X.shape[0] == len(y)
    
    n = X.shape[0]
    p = X.shape[1]
    if intercept:
        X = np.concat([np.ones(n).reshape(n,1), X], axis=1)
        X = np.matrix(X)
    else:
        X = np.matrix(X)
    
    xtx = np.matrix(X.T @ X)
    xtx_inv = xtx.getI()
    coefficients = np.array(xtx_inv @ X.T @ y).flatten()
    
    preds = np.array(X @ coefficients.T).flatten()
    residuals = np.array(y - preds)
    sigma_hat = (residuals.T @ residuals)/(n-p)
    
    def std_err(idx):
        return (np.sqrt(sigma_hat)*np.sqrt(xtx_inv[idx, idx])).item()
        # return np.sqrt(sigma_hat)*np.sqrt(xtx_inv[idx, idx])
    
    std_errs = [std_err(i) for i in range(p)]
    
    return [coefficients.tolist(), std_errs]
    # return [np.round(coefficients, 5).tolist(), np.round(std_errs, 5).tolist()]
```
